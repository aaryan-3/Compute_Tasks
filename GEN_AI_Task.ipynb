```python
import gradio as gr
import google.generativeai as genai
from groq import Groq

# -------------------------
# Configure APIs
# -------------------------

# Update with NEW keys (regenerate first!)
genai.configure(api_key="AIzaSyCtLUCRZY0xhiC4XmGi4GNqHz0T1ttgI-o")
groq_client = Groq(api_key="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")

gemini_model = genai.GenerativeModel("models/gemini-2.5-flash")


# -------------------------
# Chat Functions
# -------------------------

def chat_with_model(model_choice, message, history):
    history = history or []

    # Gemini
    if model_choice == "Gemini (2.5 Flash)":
        response = gemini_model.generate_content(message)
        bot_reply = response.text

    # Groq
    elif model_choice == "Groq (Llama 3.1 8B)":
        response = groq_client.chat.completions.create(
            model="llama-3.1-8b-instant",
            messages=[{"role": "user", "content": message}],
            temperature=0.7,
            max_tokens=300
        )
        bot_reply = response.choices[0].message.content

    else:
        bot_reply = "Invalid model."

    history.append(("You", message))
    history.append(("Bot", bot_reply))

    return history, history


# -------------------------
# Gradio UI
# -------------------------

with gr.Blocks() as demo:
    gr.Markdown("# 🤖 Multi-Chatbot UI (Gemini + Groq)")

    model_choice = gr.Dropdown(
        ["Gemini (2.5 Flash)", "Groq (Llama 3.1 8B)"],
        label="Choose your chatbot"
    )

    chatbot = gr.Chatbot(height=300)
    message = gr.Textbox(label="Enter your message")
    send = gr.Button("Send")

    send.click(
        fn=chat_with_model,
        inputs=[model_choice, message, chatbot],
        outputs=[chatbot, chatbot]
    )

demo.launch(share=False, inline=False)
```
